```python
import heapq

# Indian city road network
roads = {
    "Delhi": {
        "Agra": 230,
        "Jaipur": 280,
        "Lucknow": 550
    },

    "Agra": {
        "Delhi": 230,
        "Jaipur": 240,
        "Kanpur": 290
    },

    "Jaipur": {
        "Delhi": 280,
        "Agra": 240,
        "Ahmedabad": 670
    },

    "Lucknow": {
        "Delhi": 550,
        "Kanpur": 90,
        "Varanasi": 320
    },

    "Kanpur": {
        "Lucknow": 90,
        "Agra": 290,
        "Varanasi": 330
    },

    "Varanasi": {
        "Lucknow": 320,
        "Kanpur": 330,
        "Patna": 250
    },

    "Patna": {
        "Varanasi": 250,
        "Kolkata": 580
    },

    "Kolkata": {
        "Patna": 580,
        "Bhubaneswar": 440
    },

    "Bhubaneswar": {
        "Kolkata": 440,
        "Visakhapatnam": 440
    },

    "Visakhapatnam": {
        "Bhubaneswar": 440,
        "Hyderabad": 620
    },

    "Hyderabad": {
        "Visakhapatnam": 620,
        "Bengaluru": 570,
        "Mumbai": 710
    },

    "Bengaluru": {
        "Hyderabad": 570,
        "Chennai": 350
    },

    "Chennai": {
        "Bengaluru": 350,
        "Hyderabad": 630
    },

    "Mumbai": {
        "Hyderabad": 710,
        "Ahmedabad": 530,
        "Pune": 150
    },

    "Ahmedabad": {
        "Mumbai": 530,
        "Jaipur": 670
    },

    "Pune": {
        "Mumbai": 150,
        "Bengaluru": 840
    }
}


def shortest_route(graph, source, destination):

    distances = {city: float("inf") for city in graph}
    previous = {}

    distances[source] = 0

    queue = [(0, source)]

    while queue:

        current_cost, city = heapq.heappop(queue)

        if current_cost != distances[city]:
            continue

        if city == destination:
            break

        for next_city, road_distance in graph[city].items():

            total_cost = current_cost + road_distance

            if total_cost < distances[next_city]:

                distances[next_city] = total_cost
                previous[next_city] = city

                heapq.heappush(
                    queue,
                    (total_cost, next_city)
                )

    if distances[destination] == float("inf"):
        return None, None

    # Construct path
    route = []
    current = destination

    while current != source:
        route.append(current)
        current = previous[current]

    route.append(source)
    route.reverse()

    return route, distances[destination]


def find_city(name):

    name = name.strip().lower()

    for city in roads:
        if city.lower() == name:
            return city

    return None


start = find_city(input("Enter starting city: "))
end = find_city(input("Enter destination city: "))

if start is None or end is None:

    print("Invalid city name.")

else:

    route, distance = shortest_route(
        roads,
        start,
        end
    )

    if route:

        print("\nShortest Route:")
        print(" -> ".join(route))

        print("\nTotal Road Distance:", distance, "km")

    else:

        print("No route exists between the selected cities.")
```
